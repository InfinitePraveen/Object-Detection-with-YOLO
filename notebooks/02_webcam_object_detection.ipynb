# Optional Webcam Object Detection

This notebook demonstrates how the same YOLOv8 Nano model can process frames from a local webcam. It is optional because webcam access depends on the computer environment. CPU inference is used throughout.

```python
import cv2
from ultralytics import YOLO

model = YOLO("yolov8n.pt")
```

## Run the webcam detector

Press **q** in the OpenCV window to stop. If your computer has no webcam, skip this cell. Lower `imgsz` to 320 if CPU inference feels slow.

```python
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    raise RuntimeError("Could not open the default webcam. Check camera permissions or skip this notebook.")

while True:
    ok, frame = cap.read()
    if not ok:
        break

    results = model.predict(
        source=frame,
        imgsz=320,
        conf=0.35,
        device="cpu",
        verbose=False
    )

    annotated = results[0].plot()
    cv2.imshow("YOLOv8 Nano - Object Detection", annotated)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

## CPU tips

- Use `yolov8n.pt`, not a larger model.
- Use `imgsz=320` for faster webcam inference.
- Close other CPU-heavy applications.
- The Flask app in this repository is the recommended interview demo when webcam access is inconvenient.
